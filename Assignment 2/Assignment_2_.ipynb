### 2. Preprocess Images
   Perform the following preprocessing steps to ensure consistency and quality across the dataset:

   - ~~**Resize Images**: Resize all images to a uniform size (e.g., 256x256 pixels) to ensure consistency in the feature extraction process.~~
   
   - **Normalize Pixel Values**: Normalize pixel values to a range of `[0, 1]` by dividing pixel values by 255. This ensures consistent intensity levels for the image.
   
   - **Remove Noisy or Corrupted Images**: Filter out any corrupted or highly noisy images from the dataset, which could adversely affect feature extraction.
   
   - **Convert to RGB**: If the images are in grayscale or another color space, convert them to RGB format, as the feature extraction function expects color histograms from the Red, Green, and Blue channels.

### 3. Extract Features from Images
   - Extract the features of each image by dividing them into smaller patches (e.g., 32x32 pixels).
   - Compute the color histograms for each patch, producing a feature vector for each image.
   - Save the extracted feature vectors in `.npy` format for efficient loading during model training and evaluation.

### 4. Split Dataset into Training and Testing Sets
   - Ensure the dataset is split into separate training and testing sets before extracting features.
   - This step is crucial to prevent overfitting and ensure proper evaluation of model performance.

### 5. Create Feature Folders
   - Organize the extracted features into separate folders for the training and testing sets. For example:
     - `train/output_train_feature/`
     - `test/output_test_feature/`

### 6. Test Feature Extraction
   - Before processing the entire dataset, test the feature extraction code on a small subset of images to verify that everything works correctly.
   - This helps catch any potential errors related to image size, format, or path issues.


## Data Exploration

```python
import os
import cv2

# Function to validate images in a folder
def validate_images_in_folder(folder_path):
    image_summary = {
        'total_images': 0,
        'valid_images': 0,
        'corrupt_images': 0,
        'average_width': 0,
        'average_height': 0,
        'image_channels': {},
        'invalid_images': []
    }
    
    total_width = 0
    total_height = 0

    # Traverse through the folder
    for root, dirs, files in os.walk(folder_path):
        for file in files:
            if file.endswith(('.jpg', '.jpeg', '.png', '.bmp', '.tiff')):  # Consider common image formats
                image_path = os.path.join(root, file)
                image_summary['total_images'] += 1
                
                # Try to read the image
                try:
                    img = cv2.imread(image_path)
                    if img is None:
                        raise ValueError("Image could not be read.")
                    
                    # Update valid image count
                    image_summary['valid_images'] += 1
                    
                    # Get image properties
                    height, width, channels = img.shape
                    total_width += width
                    total_height += height
                    
                    # Track the number of channels (grayscale = 1, RGB = 3, etc.)
                    if channels not in image_summary['image_channels']:
                        image_summary['image_channels'][channels] = 0
                    image_summary['image_channels'][channels] += 1

                except Exception as e:
                    # If there's an error, treat the image as corrupt
                    image_summary['corrupt_images'] += 1
                    image_summary['invalid_images'].append((image_path, str(e)))
                    print(f"Error reading image: {image_path} - {e}")

    # Compute average width and height if there are valid images
    if image_summary['valid_images'] > 0:
        image_summary['average_width'] = total_width // image_summary['valid_images']
        image_summary['average_height'] = total_height // image_summary['valid_images']

    # Return the summary
    return image_summary

# Example usage of the validation function
def display_summary(folder_path):
    summary = validate_images_in_folder(folder_path)
    
    print("Image Validation Summary:")
    print(f"Total images found: {summary['total_images']}")
    print(f"Valid images: {summary['valid_images']}")
    print(f"Corrupt images: {summary['corrupt_images']}")
    print(f"Average image width: {summary['average_width']}")
    print(f"Average image height: {summary['average_height']}")
    print(f"Image channels distribution: {summary['image_channels']}")
    
    if summary['corrupt_images'] > 0:
        print("\nList of corrupt or invalid images:")
        for invalid_image in summary['invalid_images']:
            print(f"{invalid_image[0]} - Error: {invalid_image[1]}")


```

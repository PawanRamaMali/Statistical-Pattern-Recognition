```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
from scipy.stats import multivariate_normal

# Function to perform K-Means Clustering
def kmeans(X, k, max_iters=100):
    centroids = X[np.random.choice(X.shape[0], k, replace=False)]
    for _ in range(max_iters):
        distances = np.linalg.norm(X[:, np.newaxis] - centroids, axis=2)
        clusters = np.argmin(distances, axis=1)
        new_centroids = np.array([X[clusters == i].mean(axis=0) for i in range(k)])
        if np.all(centroids == new_centroids):
            break
        centroids = new_centroids
    return centroids, clusters

# Function to perform GMM using Expectation-Maximization (EM)
def gmm(X, k, max_iters=100, tol=1e-6):
    n, d = X.shape
    weights = np.ones(k) / k
    means = X[np.random.choice(X.shape[0], k, replace=False)]
    covariances = np.array([np.cov(X.T) for _ in range(k)])
    log_likelihoods = []

    for _ in range(max_iters):
        responsibilities = np.zeros((n, k))
        for i in range(k):
            responsibilities[:, i] = weights[i] * multivariate_normal(mean=means[i], cov=covariances[i]).pdf(X)
        responsibilities /= responsibilities.sum(axis=1, keepdims=True)

        N_k = responsibilities.sum(axis=0)
        weights = N_k / n
        means = np.dot(responsibilities.T, X) / N_k[:, np.newaxis]
        covariances = np.array([(responsibilities[:, i][:, np.newaxis] * (X - means[i])).T @ (X - means[i]) / N_k[i]
                                for i in range(k)])

        log_likelihood = np.sum(np.log(np.sum([weights[i] * multivariate_normal(mean=means[i], cov=covariances[i]).pdf(X)
                                               for i in range(k)], axis=0)))
        log_likelihoods.append(log_likelihood)

        if len(log_likelihoods) > 1 and np.abs(log_likelihoods[-1] - log_likelihoods[-2]) < tol:
            break

    return means, covariances, weights, responsibilities, log_likelihoods

# Function to plot the data and decision boundary
def plot_clusters(X, centroids, clusters, title, k):
    plt.figure(figsize=(10, 7))
    plt.scatter(X[:, 0], X[:, 1], c=clusters, cmap='viridis', marker='o', edgecolor='k')
    plt.scatter(centroids[:, 0], centroids[:, 1], color='red', marker='X', s=200, label='Centroids')
    plt.title(f'{title} with K={k}')
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.legend()
    plt.grid(True)
    plt.show()

# Function to plot log-likelihood over iterations
def plot_log_likelihoods(log_likelihoods, title, k):
    plt.figure(figsize=(10, 7))
    plt.plot(log_likelihoods, marker='o')
    plt.title(f'{title} with K={k}')
    plt.xlabel('Iterations')
    plt.ylabel('Log-Likelihood')
    plt.grid(True)
    plt.show()

# Function to evaluate the performance
def evaluate_performance(y_true, y_pred, class_names):
    cm = confusion_matrix(y_true, y_pred)
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, average='weighted')
    recall = recall_score(y_true, y_pred, average='weighted')
    f1 = f1_score(y_true, y_pred, average='weighted')

    print(f"Confusion Matrix:\n{cm}")
    print(f"Accuracy: {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall: {recall:.4f}")
    print(f"F1-Score: {f1:.4f}")

    return cm, accuracy, precision, recall, f1

# Main function to run K-Means, GMM, and evaluate the models for different values of K
def main():
    # Load the data (assuming the data is loaded as numpy arrays)
    with open('NLS_Group03.txt', 'r') as f:
        lines = f.readlines()

    lines = lines[1:]
    class1_data = np.array([line.strip().split() for line in lines[:2447]], dtype=float)
    class2_data = np.array([line.strip().split() for line in lines[2447:]], dtype=float)
    
    data = np.vstack((class1_data, class2_data))
    labels = np.array([0]*len(class1_data) + [1]*len(class2_data))

    # Values of K to try
    K = [1, 2, 4, 5, 16, 32, 64]

    # Apply K-Means and GMM for each K
    for k in K:
        print(f"\nApplying K-Means and GMM for K={k}\n")

        # Apply K-Means
        centroids, clusters = kmeans(data, k)
        plot_clusters(data, centroids, clusters, 'K-Means Clustering', k)

        # Apply GMM
        means, covariances, weights, responsibilities, log_likelihoods = gmm(data, k)
        gmm_clusters = np.argmax(responsibilities, axis=1)
        plot_clusters(data, means, gmm_clusters, 'GMM Clustering', k)
        plot_log_likelihoods(log_likelihoods, 'Log-Likelihoods during GMM', k)

        # Evaluate K-Means
        print(f"\nEvaluation for K-Means with K={k}:")
        evaluate_performance(labels, clusters, ['Class 1', 'Class 2'])

        # Evaluate GMM
        print(f"\nEvaluation for GMM with K={k}:")
        evaluate_performance(labels, gmm_clusters, ['Class 1', 'Class 2'])

if __name__ == "__main__":
    main()
```

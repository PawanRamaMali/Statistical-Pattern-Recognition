# 1.Bayes_Classifier_Using_GMM_with_PCA_Implementation

```python
import os
import numpy as np
from sklearn.decomposition import PCA
from sklearn.mixture import GaussianMixture
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import seaborn as sns

# Set environment variable to avoid memory leak warning on Windows with MKL
os.environ["OMP_NUM_THREADS"] = "1"

# Define class names globally so they can be used throughout the program
classes = ['botanical_garden', 'bus_interior', 'elevator_shaft']

# Function to create K-means model
def create_kmeans_model(data_path, n_clusters=32):
    features = []
    for class_name in classes:
        class_path = os.path.join(data_path, class_name)
        for feature_file in os.listdir(class_path):
            feature = np.load(os.path.join(class_path, feature_file))
            features.extend(feature)  # Add all 24-dimensional vectors for clustering
    features = np.array(features)
    kmeans = KMeans(n_clusters=n_clusters, random_state=0).fit(features)
    return kmeans

# Load and transform each file into a fixed 32-dimensional BoVW vector
def load_and_transform_features(data_path, kmeans):
    features = []
    labels = []
    for label, class_name in enumerate(classes):
        class_path = os.path.join(data_path, class_name)
        for feature_file in os.listdir(class_path):
            feature = np.load(os.path.join(class_path, feature_file))
            # Convert 24D vectors into a 32D BoVW vector
            cluster_labels = kmeans.predict(feature)
            bovw_vector, _ = np.histogram(cluster_labels, bins=np.arange(33))  # 32 bins
            bovw_vector = bovw_vector / len(feature)  # Normalize
            features.append(bovw_vector)
            labels.append(label)
    return np.array(features), np.array(labels)

# Create K-means model on train data
kmeans = create_kmeans_model("E:/Statistical_patter_recognition_assignment02/group03/3_class_scane_image_dataset/output_train_feature")

# Load and transform train and test features
train_features, train_labels = load_and_transform_features("E:/Statistical_patter_recognition_assignment02/group03/3_class_scane_image_dataset/output_train_feature", kmeans)
test_features, test_labels = load_and_transform_features("E:/Statistical_patter_recognition_assignment02/group03/3_class_scane_image_dataset/output_test_feature", kmeans)

# Check if features loaded correctly
if train_features.size == 0 or test_features.size == 0:
    print("Error: No features loaded for training or testing. Exiting program.")
else:
    print("Number of loaded training features:", len(train_features))
    print("Shape of train_features:", train_features.shape)

    # Step 1: Compute the covariance matrix of the training data and print it
    cov_matrix = np.cov(train_features, rowvar=False)
    print("Covariance Matrix:\n", cov_matrix)

    # Step 2: Perform eigen analysis and print eigenvalues and eigenvectors
    eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)
    sorted_indices = np.argsort(eigenvalues)[::-1]  # Sort by descending eigenvalues
    eigenvalues = eigenvalues[sorted_indices]
    eigenvectors = eigenvectors[:, sorted_indices]
    print("Eigenvalues:\n", eigenvalues)
    print("Eigenvectors (Principal Components):\n", eigenvectors)

    # Step 3: Plot eigenvalues
    plt.plot(range(1, len(eigenvalues) + 1), eigenvalues, marker='o')
    plt.xlabel("Index")
    plt.ylabel("Eigenvalue Magnitude")
    plt.title("Index vs. Eigenvalue Magnitude")
    plt.show()

    # Step 4: Apply PCA with varying reduced dimensions and GMM with different mixtures
    reduced_dimensions = [1, 2, 5, 10, 15, 20]
    mixtures = [1, 2, 4, 8]
    results = []

    for l in reduced_dimensions:
        pca = PCA(n_components=l)
        train_reduced = pca.fit_transform(train_features)
        test_reduced = pca.transform(test_features)

        for n_components in mixtures:
            # Train GMM for each class
            gmms = []
            for class_label in np.unique(train_labels):
                gmm = GaussianMixture(n_components=n_components, covariance_type='full', random_state=0)
                gmm.fit(train_reduced[train_labels == class_label])
                gmms.append(gmm)

            # Classify test data
            test_predictions = []
            for x in test_reduced:
                log_probs = [gmm.score_samples(x.reshape(1, -1)) for gmm in gmms]
                test_predictions.append(np.argmax(log_probs))

            # Calculate performance metrics
            accuracy = accuracy_score(test_labels, test_predictions)
            precision = precision_score(test_labels, test_predictions, average=None)
            recall = recall_score(test_labels, test_predictions, average=None)
            f1 = f1_score(test_labels, test_predictions, average=None)
            mean_precision = np.mean(precision)
            mean_recall = np.mean(recall)
            mean_f1 = np.mean(f1)
            conf_matrix = confusion_matrix(test_labels, test_predictions)

            # Store the results
            results.append((l, n_components, accuracy, precision, mean_precision, recall, mean_recall, f1, mean_f1))
            print(f"Reduced Dim={l}, Mixtures={n_components}, Accuracy={accuracy}")
            print(f"Precision per class: {precision}, Mean Precision: {mean_precision}")
            print(f"Recall per class: {recall}, Mean Recall: {mean_recall}")
            print(f"F1-score per class: {f1}, Mean F1-score: {mean_f1}")
            print("Confusion Matrix:\n", conf_matrix)

            # Plot confusion matrix
            plt.figure(figsize=(8, 6))
            sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues", xticklabels=classes, yticklabels=classes)
            plt.xlabel("Predicted Labels")
            plt.ylabel("True Labels")
            plt.title(f"Confusion Matrix (Dim={l}, Mixtures={n_components})")
            plt.show()

    # Display summarized results
    print("\nSummary of Results:")
    for l, n_components, accuracy, precision, mean_precision, recall, mean_recall, f1, mean_f1 in results:
        print(f"Dim: {l}, Mixtures: {n_components}, Accuracy: {accuracy:.2f}")
        print(f"Precision: {precision}, Mean Precision: {mean_precision:.2f}")
        print(f"Recall: {recall}, Mean Recall: {mean_recall:.2f}")
        print(f"F1-score: {f1}, Mean F1-score: {mean_f1:.2f}")
```

# 2.Bayes_Classifier_Using_Density_with_KNN_Implementation

```python
import os
import numpy as np
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Global class labels for consistency
classes = ['botanical_garden', 'bus_interior', 'elevator_shaft']

# Load features for training and testing
def load_features(data_path):
    features = []
    labels = []
    for label, class_name in enumerate(classes):
        class_path = os.path.join(data_path, class_name)
        for feature_file in os.listdir(class_path):
            feature = np.load(os.path.join(class_path, feature_file))
            features.append(feature.mean(axis=0))  # Aggregating features if needed
            labels.append(label)
    return np.array(features), np.array(labels)

# Load train and test features
train_features, train_labels = load_features("E:/Statistical_patter_recognition_assignment02/group03/3_class_scane_image_dataset/output_train_feature")
test_features, test_labels = load_features("E:/Statistical_patter_recognition_assignment02/group03/3_class_scane_image_dataset/output_test_feature")

# Verify data loading
if train_features.size == 0 or test_features.size == 0:
    print("Error: No features loaded for training or testing. Exiting program.")
else:
    # Define different values for K in KNN
    k_values = [1, 3, 5, 7]

    # Results container
    results = []

    for k in k_values:
        # KNN-based density estimation and Bayes classification
        test_predictions = []
        for test_point in test_features:
            densities = []
            for class_label in np.unique(train_labels):
                # Train KNN for density estimation on the subset of training data for the current class
                class_train_data = train_features[train_labels == class_label]
                knn = NearestNeighbors(n_neighbors=k)
                knn.fit(class_train_data)
                
                # Calculate the density based on the distance to the k-th nearest neighbor
                distances, _ = knn.kneighbors([test_point])
                density = 1 / (np.mean(distances) + 1e-5)  # Avoid division by zero
                densities.append(density)
            
            # Predict the class with the highest density
            predicted_class = np.argmax(densities)
            test_predictions.append(predicted_class)

        # Calculate performance metrics
        accuracy = accuracy_score(test_labels, test_predictions)
        precision = precision_score(test_labels, test_predictions, average=None)
        recall = recall_score(test_labels, test_predictions, average=None)
        f1 = f1_score(test_labels, test_predictions, average=None)
        mean_precision = np.mean(precision)
        mean_recall = np.mean(recall)
        mean_f1 = np.mean(f1)
        conf_matrix = confusion_matrix(test_labels, test_predictions)

        # Store the results
        results.append((k, accuracy, precision, mean_precision, recall, mean_recall, f1, mean_f1))
        print(f"K={k}, Accuracy={accuracy}")
        print(f"Precision per class: {precision}, Mean Precision: {mean_precision}")
        print(f"Recall per class: {recall}, Mean Recall: {mean_recall}")
        print(f"F1-score per class: {f1}, Mean F1-score: {mean_f1}")
        print("Confusion Matrix:\n", conf_matrix)

        # Plot confusion matrix
        plt.figure(figsize=(8, 6))
        sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues", xticklabels=classes, yticklabels=classes)
        plt.xlabel("Predicted Labels")
        plt.ylabel("True Labels")
        plt.title(f"Confusion Matrix (K={k})")
        plt.show()

    # Display summarized results
    print("\nSummary of Results:")
    for k, accuracy, precision, mean_precision, recall, mean_recall, f1, mean_f1 in results:
        print(f"K: {k}, Accuracy: {accuracy:.2f}")
        print(f"Precision: {precision}, Mean Precision: {mean_precision:.2f}")
        print(f"Recall: {recall}, Mean Recall: {mean_recall:.2f}")
        print(f"F1-score: {f1}, Mean F1-score: {mean_f1:.2f}")
```

# 3.Apply Fisher linear discriminant analysis (FDA)

```python
import numpy as np
import os
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
from sklearn.mixture import GaussianMixture
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import multivariate_normal

# Define class labels globally
classes = ['botanical_garden', 'bus_interior', 'elevator_shaft']

# Load features for training and testing
def load_features(data_path):
    features = []
    labels = []
    for label, class_name in enumerate(classes):
        class_path = os.path.join(data_path, class_name)
        for feature_file in os.listdir(class_path):
            feature = np.load(os.path.join(class_path, feature_file))
            features.append(feature.mean(axis=0))  # Aggregating features if needed
            labels.append(label)
    return np.array(features), np.array(labels)

# Load train and test features
train_features, train_labels = load_features("E:/Statistical_patter_recognition_assignment02/group03/3_class_scane_image_dataset/output_train_feature")
test_features, test_labels = load_features("E:/Statistical_patter_recognition_assignment02/group03/3_class_scane_image_dataset/output_test_feature")

# Verify data loading
if train_features.size == 0 or test_features.size == 0:
    print("Error: No features loaded for training or testing. Exiting program.")
else:
    # Step 1: Apply Fisher Linear Discriminant Analysis (FDA)
    lda = LDA(n_components=len(classes) - 1)  # For 3 classes, FDA projects to 2D space
    train_fda = lda.fit_transform(train_features, train_labels)
    test_fda = lda.transform(test_features)

    # Task 1: Bayes Classifier with Unimodal Gaussian
    def unimodal_gaussian_bayes(train_data, train_labels, test_data, test_labels):
        means = []
        covariances = []
        priors = []

        # Calculate mean, covariance, and prior for each class
        for class_label in np.unique(train_labels):
            class_data = train_data[train_labels == class_label]
            means.append(np.mean(class_data, axis=0))
            covariances.append(np.cov(class_data, rowvar=False))
            priors.append(len(class_data) / len(train_data))

        # Classify each test point based on likelihood under unimodal Gaussian
        test_predictions = []
        for x in test_data:
            likelihoods = [
                multivariate_normal.pdf(x, mean=means[i], cov=covariances[i]) * priors[i]
                for i in range(len(classes))
            ]
            test_predictions.append(np.argmax(likelihoods))

        # Calculate performance metrics
        accuracy = accuracy_score(test_labels, test_predictions)
        precision = precision_score(test_labels, test_predictions, average=None)
        recall = recall_score(test_labels, test_predictions, average=None)
        f1 = f1_score(test_labels, test_predictions, average=None)
        mean_precision = np.mean(precision)
        mean_recall = np.mean(recall)
        mean_f1 = np.mean(f1)
        conf_matrix = confusion_matrix(test_labels, test_predictions)

        print("\nUnimodal Gaussian Bayes Classifier Results:")
        print(f"Accuracy: {accuracy}")
        print(f"Precision per class: {precision}, Mean Precision: {mean_precision}")
        print(f"Recall per class: {recall}, Mean Recall: {mean_recall}")
        print(f"F1-score per class: {f1}, Mean F1-score: {mean_f1}")
        print("Confusion Matrix:\n", conf_matrix)

        # Plot confusion matrix
        plt.figure(figsize=(8, 6))
        sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues", xticklabels=classes, yticklabels=classes)
        plt.xlabel("Predicted Labels")
        plt.ylabel("True Labels")
        plt.title("Confusion Matrix for Unimodal Gaussian Bayes Classifier")
        plt.show()

    # Task 2: Bayes Classifier with Gaussian Mixture Model (GMM)
    def gmm_bayes(train_data, train_labels, test_data, test_labels, n_components=4):
        gmms = []
        priors = []

        # Train a GMM for each class
        for class_label in np.unique(train_labels):
            class_data = train_data[train_labels == class_label]
            gmm = GaussianMixture(n_components=n_components, covariance_type='full', random_state=0)
            gmm.fit(class_data)
            gmms.append(gmm)
            priors.append(len(class_data) / len(train_data))

        # Classify each test point based on GMM likelihood
        test_predictions = []
        for x in test_data:
            likelihoods = [
                gmm.score_samples(x.reshape(1, -1)) + np.log(priors[i])
                for i, gmm in enumerate(gmms)
            ]
            test_predictions.append(np.argmax(likelihoods))

        # Calculate performance metrics
        accuracy = accuracy_score(test_labels, test_predictions)
        precision = precision_score(test_labels, test_predictions, average=None)
        recall = recall_score(test_labels, test_predictions, average=None)
        f1 = f1_score(test_labels, test_predictions, average=None)
        mean_precision = np.mean(precision)
        mean_recall = np.mean(recall)
        mean_f1 = np.mean(f1)
        conf_matrix = confusion_matrix(test_labels, test_predictions)

        print("\nGaussian Mixture Model (GMM) Bayes Classifier Results:")
        print(f"Accuracy: {accuracy}")
        print(f"Precision per class: {precision}, Mean Precision: {mean_precision}")
        print(f"Recall per class: {recall}, Mean Recall: {mean_recall}")
        print(f"F1-score per class: {f1}, Mean F1-score: {mean_f1}")
        print("Confusion Matrix:\n", conf_matrix)

        # Plot confusion matrix
        plt.figure(figsize=(8, 6))
        sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues", xticklabels=classes, yticklabels=classes)
        plt.xlabel("Predicted Labels")
        plt.ylabel("True Labels")
        plt.title("Confusion Matrix for GMM Bayes Classifier")
        plt.show()

    # Run both classifiers
    unimodal_gaussian_bayes(train_fda, train_labels, test_fda, test_labels)
    gmm_bayes(train_fda, train_labels, test_fda, test_labels, n_components=4)
```

# 4.Perceptron-based classifier

```python
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Define the class labels for reference
classes = ['botanical_garden', 'bus_interior', 'elevator_shaft']

# Load features for training and testing
def load_features(data_path):
    features = []
    labels = []
    for label, class_name in enumerate(classes):
        class_path = os.path.join(data_path, class_name)
        for feature_file in os.listdir(class_path):
            feature = np.load(os.path.join(class_path, feature_file))
            features.append(feature.mean(axis=0))  # Aggregate features if needed
            labels.append(label)
    return np.array(features), np.array(labels)

# Perceptron Class
class Perceptron:
    def __init__(self, input_dim, learning_rate=0.1, max_epochs=1000):
        self.weights = np.zeros(input_dim + 1)  # +1 for the bias term
        self.learning_rate = learning_rate
        self.max_epochs = max_epochs

    def predict(self, x):
        activation = np.dot(x, self.weights[1:]) + self.weights[0]  # x·w + bias
        return 1 if activation >= 0 else -1

    def train(self, X, y):
        for epoch in range(self.max_epochs):
            errors = 0
            for i in range(len(X)):
                prediction = self.predict(X[i])
                if y[i] != prediction:
                    # Update rule: w = w + learning_rate * (y - prediction) * x
                    update = self.learning_rate * (y[i] - prediction)
                    self.weights[1:] += update * X[i]
                    self.weights[0] += update  # Update bias term
                    errors += 1
            if errors == 0:  # Stop early if no errors
                break

# Multi-class Perceptron-based Classifier using One-vs-Rest
class MultiClassPerceptron:
    def __init__(self, num_classes, input_dim, learning_rate=0.1, max_epochs=1000):
        self.perceptrons = [
            Perceptron(input_dim, learning_rate, max_epochs) for _ in range(num_classes)
        ]

    def train(self, X, y):
        for class_index in range(len(self.perceptrons)):
            # Convert the multi-class labels to binary labels for this class (One-vs-Rest)
            binary_y = np.where(y == class_index, 1, -1)
            self.perceptrons[class_index].train(X, binary_y)

    def predict(self, X):
        predictions = []
        for x in X:
            # Get activation scores from each perceptron
            scores = [perc.predict(x) for perc in self.perceptrons]
            predictions.append(np.argmax(scores))  # Choose the class with the highest score
        return np.array(predictions)

# Load train and test features
train_features, train_labels = load_features("E:/Statistical_patter_recognition_assignment02/group03/3_class_scane_image_dataset/output_train_feature")
test_features, test_labels = load_features("E:/Statistical_patter_recognition_assignment02/group03/3_class_scane_image_dataset/output_test_feature")

# Verify data loading
if train_features.size == 0 or test_features.size == 0:
    print("Error: No features loaded for training or testing. Exiting program.")
else:
    # Initialize and train the multi-class Perceptron classifier
    input_dim = train_features.shape[1]
    mc_perceptron = MultiClassPerceptron(num_classes=len(classes), input_dim=input_dim)
    mc_perceptron.train(train_features, train_labels)

    # Predict and evaluate on the test set
    test_predictions = mc_perceptron.predict(test_features)

    # Calculate performance metrics
    accuracy = accuracy_score(test_labels, test_predictions)
    precision = precision_score(test_labels, test_predictions, average=None)
    recall = recall_score(test_labels, test_predictions, average=None)
    f1 = f1_score(test_labels, test_predictions, average=None)
    mean_precision = np.mean(precision)
    mean_recall = np.mean(recall)
    mean_f1 = np.mean(f1)
    conf_matrix = confusion_matrix(test_labels, test_predictions)

    print("\nMulti-Class Perceptron Classifier Results:")
    print(f"Accuracy: {accuracy}")
    print(f"Precision per class: {precision}, Mean Precision: {mean_precision}")
    print(f"Recall per class: {recall}, Mean Recall: {mean_recall}")
    print(f"F1-score per class: {f1}, Mean F1-score: {mean_f1}")
    print("Confusion Matrix:\n", conf_matrix)

    # Plot confusion matrix
    plt.figure(figsize=(8, 6))
    sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues", xticklabels=classes, yticklabels=classes)
    plt.xlabel("Predicted Labels")
    plt.ylabel("True Labels")
    plt.title("Confusion Matrix for Multi-Class Perceptron Classifier")
    plt.show()
```

# 5.Logistic regression classifier 

```python
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Class labels for the dataset
classes = ['botanical_garden', 'bus_interior', 'elevator_shaft']

# Load features for training and testing
def load_features(data_path):
    features = []
    labels = []
    for label, class_name in enumerate(classes):
        class_path = os.path.join(data_path, class_name)
        for feature_file in os.listdir(class_path):
            feature = np.load(os.path.join(class_path, feature_file))
            features.append(feature.mean(axis=0))  # Aggregate features if needed
            labels.append(label)
    return np.array(features), np.array(labels)

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Logistic Regression Class
class LogisticRegression:
    def __init__(self, input_dim, learning_rate=0.1, max_epochs=1000):
        self.weights = np.zeros(input_dim + 1)  # +1 for bias
        self.learning_rate = learning_rate
        self.max_epochs = max_epochs

    def predict_proba(self, X):
        linear_combination = np.dot(X, self.weights[1:]) + self.weights[0]  # X·w + bias
        return sigmoid(linear_combination)

    def train(self, X, y):
        for epoch in range(self.max_epochs):
            predictions = self.predict_proba(X)
            errors = y - predictions
            gradient = np.dot(X.T, errors) / len(y)  # Mean gradient
            self.weights[1:] += self.learning_rate * gradient  # Update weights
            self.weights[0] += self.learning_rate * np.mean(errors)  # Update bias term

    def predict(self, X):
        return (self.predict_proba(X) >= 0.5).astype(int)  # Predict class based on probability

# Multi-Class Logistic Regression using One-vs-Rest
class MultiClassLogisticRegression:
    def __init__(self, num_classes, input_dim, learning_rate=0.1, max_epochs=1000):
        self.models = [
            LogisticRegression(input_dim, learning_rate, max_epochs) for _ in range(num_classes)
        ]

    def train(self, X, y):
        for class_index in range(len(self.models)):
            # Create binary labels for the current class
            binary_y = (y == class_index).astype(int)
            self.models[class_index].train(X, binary_y)

    def predict(self, X):
        # Predict probability for each class and choose the class with the highest probability
        class_probs = np.column_stack([model.predict_proba(X) for model in self.models])
        return np.argmax(class_probs, axis=1)

# Load train and test features
train_features, train_labels = load_features("E:/Statistical_patter_recognition_assignment02/group03/3_class_scane_image_dataset/output_train_feature")
test_features, test_labels = load_features("E:/Statistical_patter_recognition_assignment02/group03/3_class_scane_image_dataset/output_test_feature")

# Verify data loading
if train_features.size == 0 or test_features.size == 0:
    print("Error: No features loaded for training or testing. Exiting program.")
else:
    # Initialize and train the multi-class Logistic Regression classifier
    input_dim = train_features.shape[1]
    mc_logistic_regression = MultiClassLogisticRegression(num_classes=len(classes), input_dim=input_dim)
    mc_logistic_regression.train(train_features, train_labels)

    # Predict and evaluate on the test set
    test_predictions = mc_logistic_regression.predict(test_features)

    # Calculate performance metrics
    accuracy = accuracy_score(test_labels, test_predictions)
    precision = precision_score(test_labels, test_predictions, average=None)
    recall = recall_score(test_labels, test_predictions, average=None)
    f1 = f1_score(test_labels, test_predictions, average=None)
    mean_precision = np.mean(precision)
    mean_recall = np.mean(recall)
    mean_f1 = np.mean(f1)
    conf_matrix = confusion_matrix(test_labels, test_predictions)

    print("\nMulti-Class Logistic Regression Classifier Results:")
    print(f"Accuracy: {accuracy}")
    print(f"Precision per class: {precision}, Mean Precision: {mean_precision}")
    print(f"Recall per class: {recall}, Mean Recall: {mean_recall}")
    print(f"F1-score per class: {f1}, Mean F1-score: {mean_f1}")
    print("Confusion Matrix:\n", conf_matrix)

    # Plot confusion matrix
    plt.figure(figsize=(8, 6))
    sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues", xticklabels=classes, yticklabels=classes)
    plt.xlabel("Predicted Labels")
    plt.ylabel("True Labels")
    plt.title("Confusion Matrix for Multi-Class Logistic Regression Classifier")
    plt.show()
```

# 6.SVM-based classifier 

```python
import numpy as np
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Class labels for reference
classes = ['botanical_garden', 'bus_interior', 'elevator_shaft']

# Load features for training and testing
def load_features(data_path):
    features = []
    labels = []
    for label, class_name in enumerate(classes):
        class_path = os.path.join(data_path, class_name)
        for feature_file in os.listdir(class_path):
            feature = np.load(os.path.join(class_path, feature_file))
            features.append(feature.mean(axis=0))  # Aggregate features if needed
            labels.append(label)
    return np.array(features), np.array(labels)

# Load train and test features
train_features, train_labels = load_features("E:/Statistical_patter_recognition_assignment02/group03/3_class_scane_image_dataset/output_train_feature")
test_features, test_labels = load_features("E:/Statistical_patter_recognition_assignment02/group03/3_class_scane_image_dataset/output_test_feature")

# Verify data loading
if train_features.size == 0 or test_features.size == 0:
    print("Error: No features loaded for training or testing. Exiting program.")
else:
    # Define SVM configurations for each kernel type and parameter set
    svm_configs = {
        "Linear Kernel": {
            "kernel": "linear",
            "C_values": [0.1, 1, 10]
        },
        "Polynomial Kernel": {
            "kernel": "poly",
            "C_values": [0.1, 1, 10],
            "degrees": [2, 3, 4]
        },
        "RBF Kernel": {
            "kernel": "rbf",
            "C_values": [0.1, 1, 10],
            "gammas": [0.01, 0.1, 1]
        }
    }

    # Iterate over each SVM configuration
    for kernel_type, config in svm_configs.items():
        print(f"\nResults for {kernel_type}:")
        
        for C in config["C_values"]:
            if kernel_type == "Linear Kernel":
                # Train and evaluate Linear SVM
                svm = SVC(kernel=config["kernel"], C=C)
                svm.fit(train_features, train_labels)
                test_predictions = svm.predict(test_features)
                
                # Evaluate performance
                accuracy = accuracy_score(test_labels, test_predictions)
                precision = precision_score(test_labels, test_predictions, average=None)
                recall = recall_score(test_labels, test_predictions, average=None)
                f1 = f1_score(test_labels, test_predictions, average=None)
                mean_precision = np.mean(precision)
                mean_recall = np.mean(recall)
                mean_f1 = np.mean(f1)
                conf_matrix = confusion_matrix(test_labels, test_predictions)

                # Print results
                print(f"C={C}, Accuracy={accuracy}")
                print(f"Precision per class: {precision}, Mean Precision: {mean_precision}")
                print(f"Recall per class: {recall}, Mean Recall: {mean_recall}")
                print(f"F1-score per class: {f1}, Mean F1-score: {mean_f1}")
                print("Confusion Matrix:\n", conf_matrix)

            elif kernel_type == "Polynomial Kernel":
                for degree in config["degrees"]:
                    # Train and evaluate Polynomial SVM
                    svm = SVC(kernel=config["kernel"], C=C, degree=degree)
                    svm.fit(train_features, train_labels)
                    test_predictions = svm.predict(test_features)
                    
                    # Evaluate performance
                    accuracy = accuracy_score(test_labels, test_predictions)
                    precision = precision_score(test_labels, test_predictions, average=None)
                    recall = recall_score(test_labels, test_predictions, average=None)
                    f1 = f1_score(test_labels, test_predictions, average=None)
                    mean_precision = np.mean(precision)
                    mean_recall = np.mean(recall)
                    mean_f1 = np.mean(f1)
                    conf_matrix = confusion_matrix(test_labels, test_predictions)

                    # Print results
                    print(f"C={C}, Degree={degree}, Accuracy={accuracy}")
                    print(f"Precision per class: {precision}, Mean Precision: {mean_precision}")
                    print(f"Recall per class: {recall}, Mean Recall: {mean_recall}")
                    print(f"F1-score per class: {f1}, Mean F1-score: {mean_f1}")
                    print("Confusion Matrix:\n", conf_matrix)

            elif kernel_type == "RBF Kernel":
                for gamma in config["gammas"]:
                    # Train and evaluate RBF SVM
                    svm = SVC(kernel=config["kernel"], C=C, gamma=gamma)
                    svm.fit(train_features, train_labels)
                    test_predictions = svm.predict(test_features)
                    
                    # Evaluate performance
                    accuracy = accuracy_score(test_labels, test_predictions)
                    precision = precision_score(test_labels, test_predictions, average=None)
                    recall = recall_score(test_labels, test_predictions, average=None)
                    f1 = f1_score(test_labels, test_predictions, average=None)
                    mean_precision = np.mean(precision)
                    mean_recall = np.mean(recall)
                    mean_f1 = np.mean(f1)
                    conf_matrix = confusion_matrix(test_labels, test_predictions)

                    # Print results
                    print(f"C={C}, Gamma={gamma}, Accuracy={accuracy}")
                    print(f"Precision per class: {precision}, Mean Precision: {mean_precision}")
                    print(f"Recall per class: {recall}, Mean Recall: {mean_recall}")
                    print(f"F1-score per class: {f1}, Mean F1-score: {mean_f1}")
                    print("Confusion Matrix:\n", conf_matrix)

            # Plot confusion matrix for each setting
            plt.figure(figsize=(8, 6))
            sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues", xticklabels=classes, yticklabels=classes)
            plt.xlabel("Predicted Labels")
            plt.ylabel("True Labels")
            plt.title(f"Confusion Matrix for {kernel_type} (C={C})")
            if kernel_type == "Polynomial Kernel":
                plt.title(f"Confusion Matrix for {kernel_type} (C={C}, Degree={degree})")
            elif kernel_type == "RBF Kernel":
                plt.title(f"Confusion Matrix for {kernel_type} (C={C}, Gamma={gamma})")
            plt.show()
```
